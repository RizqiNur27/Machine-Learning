# 1. Import Library

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import GridSearchCV

from sklearn.decomposition import PCA
from imblearn.over_sampling import SMOTE
```

# 3. Data Cleaning

Pemeriksaan Nilai Unik pada Kolom Kategorikal

```python
print(df_train['JENIS KELAMIN'].unique())
print(df_train['STATUS MAHASISWA'].unique())
print(df_train['STATUS NIKAH'].unique())
print(df_train['STATUS KELULUSAN'].unique())
```

Menghapus Kolom yang Tidak Diperlukan

```python
df_train.drop(columns=['STATUS NIKAH'], inplace=True)
df_test.drop(columns=['STATUS NIKAH'], inplace=True)
```

Mengubah Nilai Kategorikal ke Numerik

```python
replacement = {'JENIS KELAMIN': {'LAKI - LAKI': 1, 'PEREMPUAN': 0},
                'STATUS MAHASISWA': {'MAHASISWA': 0, 'BEKERJA': 1},
                'STATUS KELULUSAN': {'TERLAMBAT': 1, 'TEPAT': 0}}

df_train = df_train.replace(replacement)
df_train.head()
```

Melakuakan Encoding pada data test

```python
df_test = df_test.replace(replacement)
df_test.head()
```

Cek Missing Value

```python
df_train.isnull().sum()
```

```python
df_test.isnull().sum()
```

Menghapus Kolom Yang Tidak Diperlukan

```python
# menghapus baris dengna nilai null pada dataset train
df_train = df_train.dropna(subset=['IPS 8'])
df_train = df_train.dropna(subset=['IPK '])

# meghapus baris dengna nilai null pada dataset test
df_test = df_test.dropna(subset=['IPS 8'])
df_test = df_test.dropna(subset=['IPK '])
```

```python
# mengcek kembali data kosong pada dataset train
df_train.isnull().sum()
```

```python
# mengcek kembali data kosong pada dataset test
df_test.isnull().sum()
```

Menghapus kolom yang tidak diperlukan

```python
df_test = df_test.drop(columns=['NAMA'])
df_test.head(1)
```

# 4. Analisis Korelasi Fitur

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Pilih hanya kolom numerik
df_train_num = df_train.select_dtypes(include=['number'])

# Menghitung matriks korelasi menggunakan spearman
corr_matrix = df_train_num.corr(method='spearman')

# Membuat heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm', cbar=True)
plt.title('Matriks Korelasi Fitur (Spearman)')
plt.show()
```

# **TAHAP MODELING**

# 6. Pemisahan data dan normalisasi

```python
# data yang dilatih
x_train = df_train.drop(columns=['STATUS KELULUSAN'])
y_train = df_train['STATUS KELULUSAN']

# data validasinya
x_test = df_test.drop(columns=['STATUS KELULUSAN'])
y_test = df_test['STATUS KELULUSAN']

# normalisasi data
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

# 7. Menangani ketidakseimbangan kelas dengan SMOTE

```python
print("before SMOTE")
print(y_train.value_counts())

sm = SMOTE(random_state=42 , k_neighbors=1)
x_train_smote, y_train_smote = sm.fit_resample(x_train_scaled, y_train)

print("after SMOTE")
print(y_train_smote.value_counts())
```

# 8. Menentukan Parameter Terbaik (Hyperparameter Tuning)

```python
from sklearn.model_selection import GridSearchCV


param_grid = {
    'n_neighbors' : list(range(1, 31)),
    'weights' : ['uniform', 'distance'],
    'metric' : ['euclidean', 'manhattan']
}

grid = GridSearchCV(KNeighborsClassifier(), param_grid, cv=5, n_jobs=-1)
grid.fit(x_train_smote, y_train_smote)

print("Best parameters: ", grid.best_params_)
print("Best score (CV): ", grid.best_score_)
```

# 9. Melatih Model KNN Terbaik dan Melakukan Prediks

```python
bes_knn = grid.best_estimator_
bes_knn.fit(x_train_smote, y_train_smote)

y_pred = bes_knn.predict(x_test_scaled)
```

# 10. Evaluasi Model dengan Classification Report

```python
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))
```

# 11. Evaluasi Model dengan Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix KNN best parameters')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()
```

# 12. Validasi Model dengan Cross-Validation

```python
from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(bes_knn, x_train_smote, y_train_smote, cv=5)
print("Cross-Validation Accuracy: ", cv_scores.mean())
print("std dev", cv_scores.std())
```

# 13. Menentukan Nilai K Optimal dengan Elbow Method

```python
errors = []
for k in range(1, 31):
    knn_temp = KNeighborsClassifier(n_neighbors=k)
    knn_temp.fit(x_train_smote, y_train_smote)
    pred = knn_temp.predict(x_test_scaled)
    errors.append(np.mean(pred != y_test))

plt.figure(figsize=(10, 6))
plt.plot(range(1, 31), errors, marker='o')
plt.title('Elbow Method KNN')
plt.xlabel('K')
plt.ylabel('Error Rate')
plt.grid(True)
plt.show()
```
